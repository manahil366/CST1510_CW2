### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`

````python
# D15
def parse_line(raw_line):
    """Turn one messy line and return a dict."""
    line = raw_line.strip()
    hostname_part, rest = line.split(",", 1)
    hostname = hostname_part.strip().lower()

    fields = {}
    for piece in rest.split(","):
        key, value = piece.split("=")
        fields[key.strip().lower()] = float(value.strip())

    return {"hostname": hostname, "used": fields["used"], "total": fields["total"]}


print(parse_line("SRV-02,used=120,total=100"))
````

### C1. Parse a whole file

Read `messy_lines.txt` (in this `LAB` folder) line by line. Parse each line with `parse_line` from D15,
and print the hostname with `OVER LIMIT` (100% or more), `WARNING` (90% or more) or `OK`.
Do not write the cleaning logic a second time.

**Expected output:**
```
srv-01 OK
srv-02 OVER LIMIT
srv-03 OK
srv-04 WARNING
```

````python
def parse_line(raw_line):
    """Turn one messy line and return a dict."""
    line = raw_line.strip()
    hostname_part, rest = line.split(",", 1)
    hostname = hostname_part.strip().lower()

    fields = {}
    for piece in rest.split(","):
        key, value = piece.split("=")
        fields[key.strip().lower()] = float(value.strip())

    return {"hostname": hostname, "used": fields["used"], "total": fields["total"]}

                                                     #parse line function

with open("messy_lines.txt", "r") as file:

    for line in file:

        record = parse_line(line)                    #read the file line by line

        hostname = record["hostname"]
        used = record["used"]
        total = record["total"]

        percent = (used / total) * 100

        if percent >= 100:
            status = "OVER LIMIT"

        elif percent >= 90:
            status = "WARNING"

        else:
            status = "OK"                        #condition

        print(hostname, status)
````

````python

"""
RECORD CHECK  -  my version
===========================

Name  :  Manahil Shaheen
Lane  :  AI
Date  :

Run it:   python template.py

Work through the numbered sections in order. Each one tells you what it must do.
Delete these instructions as you replace them with your code.
"""


# 1. FUNCTIONS
#Write parse_line(raw_line) that cleans and parses one raw, messy line
#    into a dict. Handle inconsistent spacing and case
#    Typical and above: also write status_of(...), shaped for your lane
#    (e.g. OVER LIMIT / WARNING / OK, or FLAGGED / OK). Give both a
#    one-line docstring.

def parse_line(raw_line):
    """Clean and convert a raw record into a dictionary."""

    parts = raw_line.strip().split(",")

    name = parts[0].strip()
    rating = int(parts[1].split("=")[1].strip())
    age = int(parts[2].split("=")[1].strip())

    return {"name": name, "rating": rating, "age": age}


def status_of(record):
    """Check if a record is flagged or okay."""

    if record["rating"] < 1 or record["rating"] > 5:
        return "FLAGGED"
    else:
        return "OK"



# 2. RAW RECORDS
#    AI / Data Science : "  Survey_2026 , rating=4 , age=23  "
#
#    Excellent: save these lines to a file and load them with open() /
#    a for loop over the file, instead of writing them directly here.

raw_records = [
    " Survey_2026 , rating=4 , age=23 ",
    "CUSTOMER_01, RATING=5, AGE=31",
    " survey_2025 , rating=2 , age=19 ",
    "Student_04 , rating=6 , age=21",
    " CUSTOMER_05 , rating=3 , age=25 ",
    "Survey_06, rating=bad, age=22"
]



# 3. SAVE RECORDS TO A FILE
#    Threshold : parse each record, print the cleaned dict.
#    Typical   : also call status_of(...) for each, print it alongside.
#    Excellent : print a one-line summary at the end - how many records,
#                how many flagged.

with open("records.txt", "w") as file:
    for line in raw_records:
        file.write(line + "\n")



# 4. READ AND PROCESS RECORDS
#    [ ] Run it against at least 5 genuinely messy raw records
#    [ ] Run it with a value that will not convert to a number - write the
#        error down
#    [ ] Check every function does one job, with a docstring

total = 0
flagged = 0

with open("records.txt", "r") as file:

    for line in file:

        total = total + 1

        try:
            record = parse_line(line)
            status = status_of(record)

            print("Record:", record)
            print("Status:", status)

            if status == "FLAGGED":
                flagged = flagged + 1

        except ValueError:
            print("Invalid record:", line.strip())
            print("Error: Value cannot be converted to a number.")

        print("-" * 20)


# 5. FINAL

print("FINAL SUMMARY")
print("Total records:", total)
print("Flagged records:", flagged + 1)
print("Successful records:", total - flagged - 1)
````
